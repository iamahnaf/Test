```python
df = pl.read_parquet('tcga_brca_tri_modal_dataset.parquet')
df.head()
```

```python
rna = pl.read_parquet('rna_eva_survpath_embeddings_4881.parquet')
rna.shape
```

```python
# Verify row count matches TCGA dataset (1073 rows in exact order 0..1072)
assert rna.height == df.height, f'Row mismatch: {rna.height} vs {df.height}'

# Replace numeric __index_level_0__ (or existing patient_id) with real patient_id from TCGA parquet
rna = (
    rna.drop(['__index_level_0__', 'patient_id'], strict=False)
    .with_columns(df['patient_id'])
    .select(['patient_id', pl.exclude('patient_id')])
)
rna.write_parquet('rna_eva_survpath_embeddings_4881_with_id.parquet')
print('Saved rna_eva_survpath_embeddings_4881_with_id.parquet | Shape:', rna.shape)
rna.head()
```
